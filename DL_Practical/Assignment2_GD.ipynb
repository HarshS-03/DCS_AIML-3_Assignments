```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = 15
e = 500
lr = 0.01
```

```python
x = lambda x:x**2
dx = lambda x: 2 * x
```

```python
history_X = [X]
history_loss = [x(X)]
history_gradient = [dx(X)]
```

```python
for epoch in range(e):
    gradient = dx(X)
    X = X - lr * gradient
    loss = x(X)

    history_X.append(X)
    history_loss.append(loss)
    history_gradient.append(gradient)

    print(f"Epoch {epoch+1}: Loss = {loss:.6f}")
```

```python
sns.set_theme(style="darkgrid", palette="husl")
```

```python
plt.figure(figsize=(10, 8))

x_range = [i/10 for i in range(-160, 161)]
y_range = [xi**2 for xi in x_range]

plt.plot(x_range, y_range, 'k-', linewidth=3, label='f(x) = x²', alpha=1)
epochs = list(range(len(history_X)))
scatter = plt.scatter(history_X, history_loss, c=epochs, cmap='coolwarm', s=100,
                     zorder=5, edgecolor='white', linewidth=1.5, label='Steps')
plt.plot(history_X, history_loss, 'r--', alpha=0.4, linewidth=2)

cbar = plt.colorbar(scatter)
cbar.set_label('Epoch', fontsize=11, fontweight='bold')

plt.title('Gradient Descent Path', fontsize=14, fontweight='bold')
plt.xlabel('X', fontsize=12)
plt.ylabel('f(x) = x²', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
